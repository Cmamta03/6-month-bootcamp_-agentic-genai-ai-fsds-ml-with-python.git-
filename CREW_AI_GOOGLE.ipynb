```python
writer = Agent(
    role='Generative AI & llm model developer',

    goal='How a Non Technical can improve an llm model',

    backstory="""Summerize detailed roadmap to become llm developer.
    how to land a job with more package.""",
    verbose=True,
    allow_delegation=False,
    llm=llm,
    tools=[]
)
```

```python
task1 = Task(
    description = """Summerize detailed roadmap to become llm developer.
    how to land a job with more package. Conduct a comprehensive analysis of the latest advancements in generative AI in 2024.
    Identify key trends, breakthrough technologies, and potential industry impacts, job market.
    your final answer MUST be a full analysis report""",

    agent=researcher,

    expected_output="A full analysis report on generative ai jobs & the latest advancements in AI 2024, including key trends, breckthrough technologies, and potential industry impacts."
)
```

```python
task2 = Task(
    description="""Using the insights provided, develop an engaging blog
  post that highlights the most significant AI advancements.
  Your post should be informative yet accessible, catering to a tech-savvy audience.
  Make it sound cool, avoid complex words so it doesn't sound like AI.
  Your final answer MUST be the full blog post of at least 4 paragraphs.""",

    agent=writer,
    expected_output="A full blog post about roadmap to learn llm model, datascience and highlighting the most significant AI advancements in a clear an engaging way."
)
```

```python
crew = Crew(
    agents=[researcher, writer],
    tasks=[task1, task2],
    verbose=True,
)
```
